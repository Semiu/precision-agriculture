```python
import pandas as pd
from sklearn import model_selection
```

```python
# data sample
df = pd.read_csv("../input/apple_quality.csv")
df
```

```python
# Imports the run() function from the training script

# Executes training for different combinations of:

# Models: Logistic Regression (lr), Random Forest (rf)

# Tuning methods: none (baseline), GridSearchCV, RandomizedSearchCV

import sys
sys.path.append("../scr")
from train import run
```

```python
# Models: Logistic Regression (lr), Random Forest (rf)
results = []

for model in ["lr", "rf"]:
    for fold in range(5):
        res = run(fold=fold, model_name=model, tuner="none")
        results.append(res)
```

```python
# Tuning methods: none (baseline), GridSearchCV, RandomizedSearchCV

for model in ["lr", "rf"]:
    for tuner in ["grid", "random"]:
        for fold in range(5):
            res = run(fold=fold, model_name=model, tuner=tuner)
            results.append(res)
```

```python
df_result = pd.DataFrame(results)
df_result
```

```python
#ploting comparison using accuray 
df_result.groupby(["model", "tuner"])["accuracy"].mean().plot(kind="bar")
```

```python
#ploting comparison using AUC
df_result.groupby(["model", "tuner"])["auc"].mean().plot(kind="bar")
```
